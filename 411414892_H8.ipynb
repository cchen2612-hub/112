```python
import random
for day in range(1, 4):
    user_guess = int(input(f"今天第{day}天,請輸入1-50數字"))
    lucky_number = random.randint(1, 50)
    print(f"今天的幸運數字是{lucky_number}")
    if user_guess == lucky_number:
        print("恭喜你，中獎了！")
        break
    else:
        print("你沒中獎，明天再來")
        if day < 3:
            print()
else:
    print("很遺憾3天都沒中獎，下次再試試吧！")
```

```python
import random
day=1
max_day=3
win=False
while day<=max_day and not win:
  lucky=random.randint(1,50)
  guess=int(input(f"今天第{day}天,請輸入1-50數字"))
  print(f"今天的幸運數字是{lucky}")
  if guess==lucky:
    print("恭喜你，中獎了！")
    win=True
  else:
    print("你沒中獎，明天再來")
    day+=1
if not win:
  print("很遺憾3天都沒中獎，下次再試試吧")
```
