```python
import datetime
import traceback
try:
    a = int(input("請輸入被除數: "))
    b = int(input("請輸入除數: "))
    result = a / b
    print("結果", result)
except Exception as e:
    with open("log_test.txt", "a", encoding="utf-8") as f:
        f.write(f"{datetime.datetime.now()},發生錯誤")
        f.write(traceback.format_exc())
finally:
    print("程式結束")
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def show(self):
        print("姓名", self.name)
        print("成績", self.score)
s1 = Student("小明", 90)
s2 = Student("小華", 80)
s1.show()
s2.show()
```

```python
class Bank:
    def __init__(self, name, money):
        self.name = name
        self.money = money
    def withdraw(self):
        num = int(input("請輸入提款金額: "))
        if num <= self.money:
            self.money = self.money - num
            print("帳戶名稱: ", self.name)
            print("剩餘餘額: ", self.money)
        else:
            print("餘額不足")
user = Bank("小明", 5000)
user.withdraw()
```
