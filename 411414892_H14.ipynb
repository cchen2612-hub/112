```python
def add(p,q,dis=0.05):
  total=p*q
  tax=total*dis
  total_all=total+tax
  return total,tax,total_all

p=int(input("請輸入單價:"))
q=int(input("請輸入數量:"))
total,tax,total_all=add(p,q)
print("小計:",total)
print("稅金:",tax)
print("總計:",total_all)
```

```python
def h(n):
  if n==1:
    return 1
  return n*h(n-1)
h(5)
```

```python
def h(n):
  if n==1:
    return 1
  if n==2:
    return 2
  return h(n-1)+h(n-2)
sum=h(10)
print(f"總共有{sum}種方式")
```
